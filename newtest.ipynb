```python
import pandas as pd
import random as r
```

```python
question=pd.read_csv("QuestionareDummy.csv")
```

```python
question
```

```python
score= pd.DataFrame(columns=["StudentId", "StudentName", "Score"])
score.to_csv("score.csv", index=False)
```

```python
totalQ = len(question)
randomI = r.sample(range(totalQ), 5)
print( randomI)
```

```python
studentId = int(input("Enter Your Student ID: "))
studentName = input("Enter Your Name: ")
score = pd.read_csv(score)
if studentId in list(score["StudentId"]):
    print("Student already take quiz")
else:
    print(f"\nWelcome to the Quiz, {studentName} (ID: {studentId})")
    score = 0
    selectedQuestions = []
    for i in randomI:
        selectedQuestions.append({
            "Question": question.loc[i, "Question"],
            "Option A": question.loc[i, "Option A"],
            "Option B": question.loc[i, "Option B"],
            "Option C": question.loc[i, "Option C"],
            "Option D": question.loc[i, "Option D"],
            "Right Answer": question.loc[i, "Right Answer"]})
    for j in range(len(selectedQuestions)):
        print(f"Q{j+1}: {selectedQuestions[j]['Question']}")
        print(f"A. {selectedQuestions[j]['Option A']}")
        print(f"B. {selectedQuestions[j]['Option B']}")
        print(f"C. {selectedQuestions[j]['Option C']}")
        print(f"D. {selectedQuestions[j]['Option D']}")
        answer = input("Enter your answer (A/B/C/D): ").upper()
        correctAnswer = selectedQuestions[j]["Right Answer"]
        if answer == correctAnswer:
            print("Correct")
            score = score+1
        else:
            print("Wrong Answer!")
            print(f"Correct Answer was: {correctAnswer}")
            break
    print(f"\nYour Final Score: {score}/5")
    new_score = pd.DataFrame({
        "StudentId": [studentId],
        "StudentName": [studentName],
        "Score": [score]
    })
    scoreDF = pd.concat([score, new_score])
    scoreDF.to_csv("score.csv", index=False)
    print("Score saved successfully.")
```
