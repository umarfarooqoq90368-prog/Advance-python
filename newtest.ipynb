```python
import pandas as pd
import random as r
df = pd.read_csv("QuestionareDummy.csv")
randomnumberList = []
for i in range(10):
    while True:
        random10Numbers = r.randint(0, 21)
        if random10Numbers not in randomnumberList:
            randomnumberList.append(random10Numbers)
            break
name = input("Enter your name: ")
score_df = pd.read_csv("student_score.csv")
if name in score_df["name"].values:
    print("You already attempted the quiz.")
else:
    score = 0
    for i in randomnumberList:
        print("\nQuestion:", df.iloc[i]["Question"])
        print("A:", df.iloc[i]["Option A"])
        print("B:", df.iloc[i]["Option B"])
        print("C:", df.iloc[i]["Option C"])
        print("D:", df.iloc[i]["Option D"])
        ans = input("Enter option (A/B/C/D): ").upper()
        if ans == df.iloc[i]["Right Answer"]:
            print("Correct!")
            score += 1
        else:
            print("Wrong answer. Game Over!")
            break
    print("\nYour score:", score)
    new_data = pd.DataFrame([[name, score]], columns=["name", "score"])
    print("Score saved successfully.")
```
